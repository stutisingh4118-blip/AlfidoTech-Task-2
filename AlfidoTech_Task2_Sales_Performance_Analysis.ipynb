# AlfidoTech Task 2 — Sales Performance Analysis
**Dataset:** `Task2_Superstore_Cleaned.csv`  
**Rows:** 9,800  
**Focus:** sales by region, category/sub-category, time, and product.

**Data note:** The supplied dataset contains no Profit, Quantity, Discount, or Conversion Rate field, so this notebook does not invent those metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Task2_Superstore_Cleaned.csv", parse_dates=["Order_Date","Ship_Date"])
df.head()
```

```python
print("Shape:", df.shape)
print(df.isna().sum())
total_sales=df["Sales"].sum()
orders=df["Order_ID"].nunique()
customers=df["Customer_ID"].nunique()
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Unique Orders: {orders:,}")
print(f"Unique Customers: {customers:,}")
print(f"Average Order Value: ${total_sales/orders:,.2f}")
```

## Regional Performance

```python
region_sales=df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
display(region_sales.to_frame("Sales"))
region_sales.sort_values().plot(kind="barh",figsize=(8,4.5))
plt.title("Sales by Region"); plt.xlabel("Sales"); plt.tight_layout(); plt.show()
```

## Category and Sub-Category Performance

```python
category_sales=df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
subcat_sales=df.groupby("Sub_Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales.to_frame("Sales"))
display(subcat_sales.head(10).to_frame("Sales"))
category_sales.sort_values().plot(kind="barh",figsize=(8,4.5))
plt.title("Sales by Category"); plt.xlabel("Sales"); plt.tight_layout(); plt.show()
```

## Time Trend and Seasonality

```python
df["Month"]=df["Order_Date"].dt.to_period("M").astype(str)
monthly_sales=df.groupby("Month")["Sales"].sum()
monthly_sales.plot(figsize=(10,4.5))
plt.title("Monthly Sales Trend"); plt.xlabel("Month"); plt.ylabel("Sales")
plt.xticks(rotation=45,ha="right"); plt.tight_layout(); plt.show()
print(df.assign(Year=df["Order_Date"].dt.year).groupby("Year")["Sales"].sum())
```

## Best and Lowest-Sales Products

```python
product_sales=df.groupby("Product_Name")["Sales"].sum().sort_values(ascending=False)
display(product_sales.head(10).to_frame("Sales"))
display(product_sales.tail(10).to_frame("Sales"))
product_sales.head(10).sort_values().plot(kind="barh",figsize=(9,4.5))
plt.title("Top 10 Products by Sales"); plt.xlabel("Sales"); plt.tight_layout(); plt.show()
```

## Five Tactical Improvements
1. Regional targeting — focus campaigns and inventory visibility on high-sales regions while diagnosing lower-sales regions.
2. Category-led merchandising — tailor promotions to the sales mix of Technology, Furniture and Office Supplies.
3. Sub-category bundles — create cross-sell bundles around high-sales sub-categories such as Phones and Chairs.
4. Seasonality planning — prepare inventory and campaigns before historically strong late-year months.
5. Product portfolio review — give high-sales products greater visibility and investigate persistently low-sales products for assortment, placement or pricing decisions.

These are revenue-focused recommendations because Profit/Discount/Quantity fields are not present.
